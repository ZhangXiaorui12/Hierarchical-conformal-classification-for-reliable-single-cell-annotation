```python
TS_mapping = {
    'lung ciliated cell':'ciliated cell',
    'plasmacytoid dendritic cell':'dendritic cell',
    'respiratory goblet cell':'lung goblet cell',
}
celltypes = ['B cell','basal cell','bronchial smooth muscle cell','capillary endothelial cell','ciliated cell',
             'classical monocyte','club cell','dendritic cell','effector memory CD4_positive_ alpha_beta T cell',
             'effector memory CD8_positive_ alpha_beta T cell','endothelial cell of artery',
             'endothelial cell of lymphatic vessel','fibroblast','intermediate monocyte','lung goblet cell',
             'lung multiciliated epithelial cell','macrophage','mature NK T cell','megakaryocyte',
             'mesothelial cell of pleura','myeloid leukocyte','naive thymus_derived CD4_positive_ alpha_beta T cell',
             'naive thymus_derived CD8_positive_ alpha_beta T cell','natural killer cell','non_classical monocyte',
             'pericyte','plasma cell','pulmonary alveolar type 1 cell','pulmonary alveolar type 2 cell',
             'pulmonary ionocyte','tracheobronchial serous cell','vascular associated smooth muscle cell',
             'vein endothelial cell',]
```

```python
TS.obs['cell_type'] = TS.obs['cell_type'].str.replace('-', '_', regex=False)
TS.obs['cell_type'] = TS.obs['cell_type'].str.replace(',', '_', regex=False)
TS.obs['cell_type'] = TS.obs['cell_type'].replace(TS_mapping)
if celltypes:
    TS = TS[TS.obs['cell_type'].isin(celltypes)]
```

```python
TS.write('TS.h5ad')
```
