```python
#Assign query data and reference data
from sklearn.model_selection import train_test_split
import numpy as np

indices = all_adata.obs_names
labels = all_adata.obs['cell_type']

ref_idx, tmp_idx = train_test_split(
    indices, 
    test_size=0.2,          
    stratify=labels,
    random_state=42
)
labels_tmp = labels[tmp_idx]

query_idx, calib_idx = train_test_split(
    tmp_idx, 
    test_size=0.2,          
    stratify=labels_tmp,
    random_state=42
)

all_adata.obs['split'] = 'unassigned'
all_adata.obs.loc[ref_idx, 'split'] = 'train'
all_adata.obs.loc[query_idx, 'split'] = 'test'
all_adata.obs.loc[calib_idx, 'split'] = 'calibrate'

print('train:', len(ref_idx), 'test:', len(query_idx), 'calibrate:', len(calib_idx))

all_adata.write(filtered_data_file)
```

```python
ref_idx = processed_adata.ref_adata.obs_names
query_idx = processed_adata.query_adata.obs_names

#Convert celltype to serial number
sorted_types = sorted(processed_adata.ref_adata.obs['cell_type'].unique())
mapping = {cell_type: i for i, cell_type in enumerate(sorted_types)}
print(mapping)
cell_type_id = processed_adata.query_adata.obs['cell_type'].replace(mapping)
cell_type_id = cell_type_id.astype(int)

#Save the predicted probabilities
for result_key in list(adata.uns["prediction_keys"]):
    probability_matrix = adata.uns[f'{result_key}_probability_matrix'].copy().loc[query_idx,]
    output_df = pd.concat([cell_type_id, probability_matrix], axis=1)
    output_df.columns = ['label'] + [f'label_{i}' for i in probability_matrix.columns]
    output_df.insert(0, 'split', adata[output_df.index].obs['split'])
    output_df.insert(0, 'id', range(len(output_df)))
    os.makedirs(f'./{probability_matrix_path}', exist_ok=True)
    output_df.to_csv(f'./{probability_matrix_path}/{result_key}.csv', index=False)
```

```python
import pandas as pd
import numpy as np
classifier_file = "output/LCA_TS/probability_matrix/popv_celltypist_prediction.csv"
classifier_df = pd.read_csv(classifier_file)
classifier_df1 = classifier_df.loc[classifier_df['split']=='calibrate',:]
prob_matrix1 = classifier_df1.iloc[:,3:]
label1 = classifier_df1['label']
pred1 = np.argmax(prob_matrix1,axis=1)
print('Calibrate Accuracy:', sum(pred1 == label1)/len(label1))
classifier_df2 = classifier_df.loc[classifier_df['split']=='test',:]
prob_matrix2 = classifier_df2.iloc[:,3:]
label2 = classifier_df2['label']
pred2 = np.argmax(prob_matrix2,axis=1)
print('Test Accuracy:', sum(pred2 == label2)/len(label2))
```
